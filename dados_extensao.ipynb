#LIMPEZA DOS DADOS

```python
import pandas as pd 

df = pd.read_csv(r'atividades_de_extensao.csv')

print(df.columns)

print('\n')

display(df.head(10))


situacao_contagem = df['Situacao'].value_counts()
print('\n')

print(situacao_contagem)

# Lista de situações que são "ruído" e devem sair do DataFrame
situacoes_para_remover = [
    'REMOVIDA', 
    'CADASTRO EM ANDAMENTO', 
    'TEMPO DE CADASTRO EXPIRADO'
]

# Filtra mantendo apenas as linhas cuja situação NÃO ESTÁ (~) na lista de remoção
df_painel = df[~df['Situacao'].isin(situacoes_para_remover)]

print('\n')

# Verifica como ficou o novo tamanho do DataFrame
print(f"\nLinhas antes: {len(df)} | Linhas depois: {len(df_painel)}")

print('\n')

df_painel.info()
```

CRIAÇÃO DO AGRUPAMENTO DAS SITUAÇÕES

🟢 Concluídos: CONCLUÍDA, PENDENTE DE RELATÓRIO

🔵 Ativos: EM EXECUÇÃO, APROVADO SEM RECURSOS, APROVADO COM RECURSOS

🔴 Cancelados / Reprovados: REPROVADA, PROJETO CANCELADO, DESCLASSIFICADO, NÃO APROVADO PELOS DEPARTAMENTOS

🟡 Em Trâmite (Funil): Todas as outras que sobraram (AGUARDANDO APROVAÇÃO DOS DEPARTAMENTOS, PROPOSTA DEVOLVIDA PARA COORDENADOR REEDITAR, AGUARDANDO AVALIAÇÃO', SUBMETIDA, AVALIAÇÃO REALIZADA, ANALISANDO SOLICITAÇÃO DE RECONSIDERAÇÃO)

```python
import pandas as pd

# 2. Criamos as listas para as nossas categorias principais
concluidos = ['CONCLUÍDA', 'PENDENTE DE RELATÓRIO']
ativos = ['EM EXECUÇÃO', 'APROVADO SEM RECURSOS', 'APROVADO COM RECURSOS']
falhas = ['REPROVADA', 'PROJETO CANCELADO', 'DESCLASSIFICADO', 'NÃO APROVADO PELOS DEPARTAMENTOS']

# 3. Criamos uma função simples para classificar cada linha
def classificar_status(situacao):
    if situacao in concluidos:
        return 'Concluído'
    elif situacao in ativos:
        return 'Ativo'
    elif situacao in falhas:
        return 'Cancelado/Reprovado'
    else:
        return 'Em Trâmite' # O que não caiu nas listas acima, é porque está rodando no sistema

# 4. Criamos a nova coluna aplicando a regra
df_painel['Status_macro'] = df_painel['Situacao'].apply(classificar_status)

display(df_painel.head(10))
```

Transformação das colunas de datas em datas

```python
df_painel.info()

# Lista com as colunas que precisam virar data
colunas_de_data = ['Data Inicio', 'Data Fim', 'Data Cadastro']

# Loop corrigido: deixamos o Pandas identificar o padrão YYYY-MM-DD naturalmente
for coluna in colunas_de_data:
    df_painel[coluna] = pd.to_datetime(df_painel[coluna], errors='coerce')

# Verifique o resultado
df_painel.info()

display(df_painel)
```

CRIAR TIPOS DE AÇÕES

```python
# 1. Criamos um dicionário com as regras que você definiu
dicionario_acoes = {
    'EV': 'Evento',
    'CR': 'Curso',
    'PD': 'Produto',
    'PG': 'Programa',
    'PJ': 'Projeto'
}

# 2. Pegamos apenas as 2 primeiras letras da coluna 'Codigo'
# O .str[:2] significa: "pegue o texto do início até o segundo caractere"
letras_iniciais = df_painel['Codigo'].str[:2]

# 3. Mapeamos essas letras para os nomes completos e criamos a nova coluna
df_painel['Tipo de Ação'] = letras_iniciais.map(dicionario_acoes)

# 4. (Opcional de Segurança) Caso exista algum código fora desse padrão na base, 
# ele ficaria nulo. Podemos preencher com "OUTROS" para não estragar o painel:
df_painel['Tipo de Ação'] = df_painel['Tipo de Ação'].fillna('OUTROS')

# 5. Verificamos o resultado da nossa nova categorização
print("Quantidade de ações por tipo:")
print(df_painel['Tipo de Ação'].value_counts())

display(df_painel)

```

```python
import pandas as pd
import re

# 1. Dicionário com as siglas e os nomes exatamente como você pediu
dicionario_centros = {
    'CBIOTEC': 'CBiotec - Centro de Biotecnologia',
    'CCEN': 'CCEN - Centro de Ciências Exatas',
    'CCHLA': 'CCHLA - Centro de Ciências Humanas e Letras',
    'CCJ': 'CCJ - Centro de Ciências Jurídicas',
    'CCM': 'CCM - Centro de Ciências Médicas',
    'CCS': 'CCS - Centro de Ciências da Saúde',
    'CCSA': 'CCSA - Centro de Ciências Sociais Aplicadas',
    'CCTA': 'CCTA - Centro de Comunicação e Artes',
    'CEAR': 'CEAR - Centro de Energias Renováveis',
    'CE': 'CE - Centro de Educação',
    'CI': 'CI - Centro de Informática',
    'CTDR': 'CTDR - Centro de Tecnologia e Desenvolvimento Regional',
    'CT': 'CT - Centro de Tecnologia',
    'CCA': 'CCA - Centro de Ciências Agrárias',
    'CCHSA': 'CCHSA - Centro de Ciências Humanas, Sociais e Agrárias',
    'CCAE': 'CCAE - Centro de Ciências Aplicadas e Educação'
}

# 2. Função inteligente para achar a sigla exata (evitando que "CI" confunda com "CIÊNCIAS")
def extrair_centro(unidade_texto):
    texto = str(unidade_texto).upper()
    
    # Testa cada sigla do nosso dicionário
    for sigla, nome_formatado in dicionario_centros.items():
        # O \b significa "fronteira de palavra". Ele garante que só vai achar a sigla isolada.
        if re.search(rf'\b{sigla}\b', texto):
            return nome_formatado
            
    # Se não achar nenhuma sigla da UFPB, retorna o conteúdo original da unidade
    return unidade_texto

# 3. Aplicando a função para criar a nova coluna
df_painel['Centro de Ensino'] = df_painel['Unidade'].apply(extrair_centro)

#ajustar a ordem das colunas

ordem_colunas = [
    'Ano Projeto', 
    'Codigo', 
    'Tipo de Ação', 
    'Titulo', 
    'Centro de Ensino',
    'Unidade', 
    'Data Inicio', 
    'Data Fim',  
    'Situacao', 
    'Status_macro',  
    'Fonte Financiamento'
]

# 3. Aplicar a nova ordem (isso também vai ocultar colunas indesejadas que não estão na lista)
df_painel = df_painel[ordem_colunas]

# 4. Restringindo a Análise Histórica (Apenas de 2020 a 2025)
# Mantém apenas os projetos onde o ano é maior ou igual a 2020 E menor ou igual a 2025
df_historico = df_painel[(df_painel['Ano Projeto'] >= 2020) & (df_painel['Ano Projeto'] <= 2025)].copy()

# 5. Sobrescrevendo o arquivo CSV do Painel com a base nova
caminho_exato = r'G:\Meu Drive\Curso_hastag\Projeto Integrador II\base_painel_2020_2025.csv'
df_historico.to_csv(caminho_exato, index=False, encoding='utf-8')

print("Transformações concluídas!")
print(f"Total de projetos filtrados (2020 a 2025): {len(df_historico)}")
print("A coluna 'CENTRO DE ENSINO' foi criada e o CSV foi atualizado.")

display(df_historico)
```
